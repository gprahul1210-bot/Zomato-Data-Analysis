## 🧹 Data Cleaning

Real-world data is rarely analysis-ready. Here we:
- Check for missing values
- Check for duplicate rows
- Verify/adjust data types
- Handle missing `Cuisines` values
- Create a clean "primary cuisine" column
- Convert Yes/No text columns into consistent, readable categorical values


```python
# 1. Check for missing values in every column
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0])
```

**Finding:** Only the `Cuisines` column has missing values (a handful of rows). All other
columns are complete.

```python
# 2. Check for duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")

# Remove duplicates if any exist (safe to run even if there are none)
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")
```

```python
# 3. Verify data types are appropriate for analysis
print(df.dtypes)
```

```python
# 4. Handle missing Cuisines values
# We use try-except to safely fill missing values, guarding against unexpected errors
try:
    df['Cuisines'] = df['Cuisines'].fillna('Not Specified')
    print("Missing Cuisines values handled successfully.")
except Exception as e:
    print(f"Something went wrong while cleaning Cuisines: {e}")

# Confirm no missing values remain
print("\nRemaining missing values:")
print(df.isnull().sum().sum())
```

```python
# 5. Extract the PRIMARY cuisine (first cuisine listed) for each restaurant
# Using a LIST COMPREHENSION to build a new list of primary cuisines
primary_cuisines = [str(c).split('|')[0].strip() for c in df['Cuisines']]

# Add it back to the dataframe as a new column
df['Primary_Cuisine'] = primary_cuisines

df[['RestaurantName', 'Cuisines', 'Primary_Cuisine']].head()
```

```python
# 6. Standardize Yes/No columns into a more descriptive category using a FUNCTION
def yes_no_label(value):
    """Convert a Yes/No string into a clean, human-readable label.
    Demonstrates a function with a parameter and a return value.
    """
    if value == 'Yes':
        return 'Available'
    elif value == 'No':
        return 'Not Available'
    else:
        return 'Unknown'

# Apply the function to the relevant service columns
df['Online_Delivery_Status'] = df['Has_Online_delivery'].apply(yes_no_label)
df['Table_Booking_Status'] = df['Has_Table_booking'].apply(yes_no_label)

df[['Has_Online_delivery', 'Online_Delivery_Status',
    'Has_Table_booking', 'Table_Booking_Status']].head()
```

```python
# 7. Categorize ratings into readable buckets using IF / ELIF / ELSE
def rating_category(rating):
    """Classify a numeric rating into a business-friendly category."""
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Very Good'
    elif rating >= 3.0:
        return 'Good'
    elif rating >= 2.0:
        return 'Average'
    else:
        return 'Poor'

df['Rating_Category'] = df['Rating'].apply(rating_category)

# Quick check of the distribution of categories created
df['Rating_Category'].value_counts()
```

```python
# 8. Label Price_range (1-4) into readable tiers using a LAMBDA FUNCTION
price_labels = {1: 'Low', 2: 'Medium', 3: 'High', 4: 'Very High'}

# Lambda used for a short, inline transformation
df['Price_Tier'] = df['Price_range'].apply(lambda p: price_labels.get(p, 'Unknown'))

df[['Price_range', 'Price_Tier']].drop_duplicates().sort_values('Price_range')
```

### ✅ Data Cleaning Summary
- Missing `Cuisines` values filled with `'Not Specified'`
- Duplicate rows removed (if any were present)
- New engineered columns added: `Primary_Cuisine`, `Rating_Category`, `Price_Tier`,
  `Online_Delivery_Status`, `Table_Booking_Status`
- All columns now have consistent, analysis-ready data types

The dataset is now clean and ready for exploratory analysis.

## 💼 Business Questions

Now we answer concrete questions a restaurant business or Zomato stakeholder would care about,
using functions, loops, dictionaries, and safe error handling.


```python
# Q1: Which city has the highest number of listed restaurants?
top_city = df['City'].value_counts().idxmax()
top_city_count = df['City'].value_counts().max()
print(f"Q1: '{top_city}' has the most listed restaurants, with {top_city_count} restaurants.")
```

```python
# Q2: What are the top 5 cities by AVERAGE rating (minimum 20 restaurants, for reliability)?
city_stats = df.groupby('City').agg(
    restaurant_count=('RestaurantID', 'count'),
    avg_rating=('Rating', 'mean')
)

# Filter to cities with a meaningful number of restaurants, then sort
reliable_cities = city_stats[city_stats['restaurant_count'] >= 20]
top_rated_cities = reliable_cities.sort_values('avg_rating', ascending=False).head(5)

print("Q2: Top 5 cities by average rating (min. 20 restaurants):")
top_rated_cities.round(2)
```

```python
# Q3: What are the top 10 most popular cuisines overall?
# Using a FOR LOOP + DICTIONARY to manually count cuisine occurrences
# (each restaurant can serve multiple cuisines, split by '|')

cuisine_counts = {}   # empty dictionary to store cuisine -> count

for cuisine_string in df['Cuisines']:
    # Split the pipe-separated cuisines into a clean list
    cuisines_list = [c.strip() for c in str(cuisine_string).split('|')]
    for cuisine in cuisines_list:
        if cuisine in cuisine_counts:
            cuisine_counts[cuisine] += 1
        else:
            cuisine_counts[cuisine] = 1

# Sort the dictionary by count and take the top 10
top_10_cuisines = sorted(cuisine_counts.items(), key=lambda x: x[1], reverse=True)[:10]

print("Q3: Top 10 most popular cuisines (restaurants may serve more than one):")
for rank, (cuisine, count) in enumerate(top_10_cuisines, start=1):
    print(f"{rank:2d}. {cuisine:<20} — {count} restaurants")
```

```python
# Q4: Does offering Online Delivery correlate with a higher average rating?
delivery_rating = df.groupby('Online_Delivery_Status')['Rating'].mean().round(2)
print("Q4: Average rating by online delivery availability:")
print(delivery_rating)

diff = delivery_rating.get('Available', 0) - delivery_rating.get('Not Available', 0)
if diff > 0:
    print(f"\n=> Restaurants WITH online delivery rate ~{diff:.2f} points higher on average.")
elif diff < 0:
    print(f"\n=> Restaurants WITHOUT online delivery rate ~{abs(diff):.2f} points higher on average.")
else:
    print("\n=> No meaningful difference in average rating.")
```

```python
# Q5: Does Table Booking availability affect average cost for two?
booking_cost = df.groupby('Table_Booking_Status')['Average_Cost_for_two'].mean().round(0)
print("Q5: Average cost for two, by table booking availability:")
print(booking_cost)
```

```python
# Q6: What percentage of restaurants offer online delivery?
delivery_percentage = (df['Has_Online_delivery'] == 'Yes').mean() * 100
print(f"Q6: {delivery_percentage:.1f}% of restaurants offer online delivery.")
```

```python
# Q7: Which price range receives the most user engagement (votes)?
votes_by_price = df.groupby('Price_Tier')['Votes'].mean().round(0).reindex(price_order)
print("Q7: Average votes received, by price tier:")
print(votes_by_price)
```

```python
# Q8: Top 10 most-voted (most popular) restaurants
top_voted = df.sort_values('Votes', ascending=False)[
    ['RestaurantName', 'City', 'Votes', 'Rating']
].head(10).reset_index(drop=True)

print("Q8: Top 10 most-voted restaurants:")
top_voted
```

```python
# Q9: Build a reusable FUNCTION to summarize any city's stats on demand
def city_summary(city_name):
    """
    Return a dictionary summarizing key stats for a given city.
    Demonstrates: function with parameter + return, dictionary, try-except.
    """
    try:
        city_df = df[df['City'] == city_name]
        if city_df.empty:
            raise ValueError(f"No restaurants found for city '{city_name}'.")

        summary = {
            'city': city_name,
            'restaurant_count': int(city_df.shape[0]),
            'avg_rating': round(city_df['Rating'].mean(), 2),
            'avg_cost_for_two': round(city_df['Average_Cost_for_two'].mean(), 0),
            'online_delivery_pct': round((city_df['Has_Online_delivery'] == 'Yes').mean() * 100, 1)
        }
        return summary

    except ValueError as e:
        # Graceful handling for a city that doesn't exist in the data
        print(f"Warning: {e}")
        return None

# Q9: Try it on a real city, and on a made-up one to show safe error handling
print("Q9: City summary example (New Delhi):")
print(city_summary('New Delhi'))

print("\nSafe handling of an unknown city:")
print(city_summary('Atlantis'))
```

```python
# Q10: Which cities have the highest average cost for two? (Top 10, min. 10 restaurants)
cost_stats = df.groupby('City').agg(
    restaurant_count=('RestaurantID', 'count'),
    avg_cost_for_two=('Average_Cost_for_two', 'mean')
)
reliable_cost_cities = cost_stats[cost_stats['restaurant_count'] >= 10]
priciest_cities = reliable_cost_cities.sort_values('avg_cost_for_two', ascending=False).head(10)

print("Q10: Top 10 priciest cities (avg. cost for two, min. 10 restaurants):")
priciest_cities.round(0)
```

## 💡 Business Recommendations

1. **Expand into under-represented cities.** Since restaurant listings are concentrated in a
   few cities, Zomato/restaurant chains have a growth opportunity in cities with fewer current
   listings but reasonable population/demand.
2. **Help low-rated, low-price restaurants improve.** Since rating variance is highest in the
   "Low" price tier, targeted support (menu, hygiene, service tools) could raise ratings and
   customer trust for budget restaurants — a large segment of the platform.
3. **Promote online delivery adoption in lagging cities.** Cities with low online delivery
   penetration represent an opportunity — enabling delivery could unlock incremental orders and
   revenue for local restaurants.
4. **Double down on high-performing cuisines,** while also using data to identify underserved
   or emerging cuisine categories worth promoting to diversify the catalog.
5. **Use "Votes" as an engagement signal, not just a popularity metric.** Since votes rise with
   price tier, engagement campaigns (reviews, offers) may be especially effective when targeted
   at mid-to-high price tier restaurants to convert visits into visible social proof.
6. **Encourage table booking adoption among quality mid-tier restaurants** to help them capture
   the customer experience benefits currently concentrated in pricier restaurants.
